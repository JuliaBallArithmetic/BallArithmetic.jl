```julia
import Pkg; Pkg.add("RigorousInvariantMeasures")
```

```julia
using RigorousInvariantMeasures
```

```julia
B = FourierAnalytic(128, 16384)
```

```julia
T(x) = 3.83*x*(1-x)
```

```julia
NK = GaussianNoise(B, 0.1)
```

```julia
P = DiscretizedOperator(B, T; max_iter = 100, ϵ = 0.00000000000001)
```

```julia
using IntervalArithmetic
```

```julia
midI = IntervalArithmetic.mid
```

```julia
Pfloat = midI.(real.(P))+im*midI.(imag.(P))
```

```julia
Q = NK.NK*Pfloat
```

```julia
using Pseudospectra, Plots
```

```julia
spectralportrait(Q)
```

```julia
plot!([cos(t) for t in 0:0.01:2π], [sin(t) for t in 0:0.01:2π], label = "")
```
